```python
# 올바른 HuggingFaceHub 설정
import os
from dotenv import load_dotenv
from langchain_community.llms import HuggingFaceHub

load_dotenv()

# API 토큰 확인
api_token = os.getenv("HUGGINGFACEHUB_API_TOKEN")
if not api_token:
    print("⚠️ HUGGINGFACEHUB_API_TOKEN이 설정되지 않았습니다!")
    print("해결 방법:")
    print("1. https://huggingface.co/settings/tokens 에서 토큰 생성")
    print("2. .env 파일에 추가: HUGGINGFACEHUB_API_TOKEN=your_token_here")
    print("\n또는 무료 대안인 Ollama를 사용하세요 (아래 셀 참조)")
else:
    print("✅ API 토큰이 설정되었습니다")

# 안정적인 모델로 LLM 초기화
llm = HuggingFaceHub(
    repo_id="google/flan-t5-base",
    model_kwargs={
        "temperature": 0.2,
        "max_length": 200
    }
)
```

```python
# HTML 생성 테스트
prompt = """
Generate a short html code to a simple webpage with a header, a subheader ,and a text body.
<!DOCTYPE html>
<html>
"""

try:
    if api_token:
        print("HTML 생성 중...")
        response = llm.invoke(prompt)
        print("생성된 HTML:")
        print(response)
    else:
        print("API 토큰이 없어서 실행할 수 없습니다.")
except Exception as e:
    print(f"❌ 오류 발생: {e}")
    print("\n가능한 해결책:")
    print("1. API 토큰을 확인하세요")
    print("2. 인터넷 연결을 확인하세요")
    print("3. 다른 모델을 시도해보세요 (아래 참조)")
```

```python
# 대안 1: Ollama 사용 (로컬 실행, API 토큰 불필요)
from langchain_community.llms import Ollama

# Ollama 설치 방법:
# 1. https://ollama.ai 에서 Ollama 다운로드
# 2. 터미널에서: ollama pull llama2
# 3. 또는: ollama pull mistral

try:
    ollama_llm = Ollama(model="llama2")  # 또는 "mistral"
    response = ollama_llm.invoke(prompt)
    print("🎉 Ollama로 생성된 HTML:")
    print(response)
except Exception as e:
    print(f"Ollama를 사용할 수 없습니다: {e}")
    print("Ollama가 설치되어 있고 실행 중인지 확인하세요.")
```

```python
# .env 파일 생성 도우미
import os
from pathlib import Path

# Jupyter 노트북에서는 __file__이 없으므로 현재 디렉토리 사용
current_dir = os.getcwd()
env_path = os.path.join(current_dir, ".env")
env_exists = os.path.exists(env_path)

print(f"현재 디렉토리: {current_dir}")
print(f".env 파일 경로: {env_path}")

if not env_exists:
    print("\n📝 .env 파일이 없습니다. 생성 예시:")
    print("\n# .env 파일 내용:")
    print("HUGGINGFACEHUB_API_TOKEN=your_token_here")
    print("OPENAI_API_KEY=your_openai_key_here")
    print("\n파일을 생성하려면 터미널에서:")
    print(f"echo 'HUGGINGFACEHUB_API_TOKEN=your_token_here' > {env_path}")
else:
    print("\n✅ .env 파일이 존재합니다")
    
    # 토큰이 설정되어 있는지 확인
    try:
        with open(env_path, 'r') as f:
            content = f.read()
            if 'HUGGINGFACEHUB_API_TOKEN' in content:
                print("✅ HUGGINGFACEHUB_API_TOKEN이 .env에 있습니다")
                # 토큰 값 일부를 가려서 표시
                import re
                match = re.search(r'HUGGINGFACEHUB_API_TOKEN=(.+)', content)
                if match:
                    token = match.group(1).strip()
                    if len(token) > 10:
                        masked_token = token[:7] + "..." + token[-4:]
                        print(f"   토큰: {masked_token}")
            else:
                print("⚠️  HUGGINGFACEHUB_API_TOKEN을 .env 파일에 추가하세요")
    except Exception as e:
        print(f"⚠️  .env 파일 읽기 오류: {e}")
```

```python
# 디버깅: 더 간단한 모델로 테스트
# StopIteration 오류가 계속 발생하면 이 셀을 실행하세요

# 옵션 1: 다른 T5 모델 시도
try:
    simple_llm = HuggingFaceHub(
        repo_id="google/flan-t5-small",  # 더 작은 모델
        model_kwargs={"temperature": 0.1, "max_length": 50}
    )
    response = simple_llm.invoke("Translate to French: Hello world")
    print("T5 Small 응답:", response)
except Exception as e:
    print(f"T5 Small 오류: {e}")

# 옵션 2: 텍스트 생성 전용 모델
try:
    gpt2_llm = HuggingFaceHub(
        repo_id="gpt2",  # 기본 GPT-2 모델
        model_kwargs={"temperature": 0.7, "max_length": 50}
    )
    response = gpt2_llm.invoke("Once upon a time")
    print("\nGPT-2 응답:", response)
except Exception as e:
    print(f"GPT-2 오류: {e}")
```
