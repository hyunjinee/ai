```python

from dotenv import load_dotenv
from pydantic_ai import Agent

load_dotenv()

basic_agent = Agent(
  # OpenAI 모델
  model = "ollama:gemma:7b", # 빠르고 저렴한 모델
  # model = "gpt-4o-mini", # 비용 효율적인 모델
  # model = "gpt-3.5-turbo", # 빠르고 저렴한 모델

  # Anthropic 모델 (Claude API 키 필요)
  # model = "claude-3-opus-20240229", # 최고 성능 모델
  # model = "claude-3-sonnet-20240229", # 균형 잡힌 모델
  # model = "claude-3-haiku-20240307", # 빠른 응답 모델

  # Ollama 모델 (로컬 설치 필요)
  # model = "ollama://llama3", # Meta의 Llama 3 모델
  # model = "ollama://mistral", # Mistral AI 모델
  system_prompt= "You are a helpful e-commerce support agent. Be consice and friendly."
)

prompt = "Say hello to the customer and ask how you can help with their order today."

print("Synchronous response:")
result = basic_agent.run_sync(prompt)
print(result.data)
```
