벡터 검색을 위해서는 먼저 인덱스를 만들어야한다.
인덱스는 관계형 데이터베이스의 테이블과 비슷한 레벨로 볼 수 있다.
인덱스에 벡터를 저장하는데, 이렇게 벡터를 저장하는 과정을 색인한다라고 표현한다.

색인단계에서는 인덱스의 메모리 사용량과 색인 시간이 가장 중요하고 검색 단계에서는 검색 시간과 재현율이 중요하다. 재현율이란, 실제로 가장 가까운 K개의 정답 데이터 중 몇 개가 검색 결과로 반환됐는지 그 비율을 나타낸 값이다. 

KNN은 모든 벡터와 거리를 측정해 결과를 반환하기 때문에 재현율은 100%이다. 

아래 KNN 검색 실습에서는 데이터가 늘어날 때 메모리 사용량과 색인시간, 검색 시간이 어떻게 변하는지를 중심으로 알아본다. 

```python
import psutil

def get_memory_usage_mb():
  process = psutil.Process()
  memory_info = process.memory_info()
  return memory_info.rss / (1024 * 1024)
```
